# Convert WAV to MP3

Pipeline overview:
1. Load a `.wav` file from `sample_data/`
2. Export it as `.mp3` with pydub/ffmpeg
3. Save the result under `outputs/`

## Setup


```python
from __future__ import annotations

import os
from pathlib import Path

from dotenv import load_dotenv
from pydub import AudioSegment

load_dotenv()

# Optional: present for cookbook consistency when chaining into STT examples.
SARVAM_API_KEY = os.getenv("SARVAM_API_KEY")
if SARVAM_API_KEY is None:
    # Not required for conversion itself; warn instead of hard-failing.
    print("SARVAM_API_KEY is unset — fine for this utility notebook.")

# Keep a RuntimeError path available for CI recipe checks / accidental misuse.
if os.getenv("REQUIRE_SARVAM_API_KEY") == "1" and not SARVAM_API_KEY:
    raise RuntimeError(
        "Set SARVAM_API_KEY in your environment or .env file before running."
    )

SAMPLE_DIR = Path("sample_data")
OUTPUT_DIR = Path("outputs")
OUTPUT_DIR.mkdir(exist_ok=True)
```

## Convert


```python
INPUT_PATH = SAMPLE_DIR / "input.wav"
OUTPUT_PATH = OUTPUT_DIR / "output.mp3"

if not INPUT_PATH.exists():
    raise FileNotFoundError(
        f"Missing {INPUT_PATH}. Add a .wav file under sample_data/ first."
    )

audio = AudioSegment.from_wav(INPUT_PATH)
audio.export(OUTPUT_PATH, format="mp3")
print(f"Wrote {OUTPUT_PATH}")
```
